```python
# Using a break statement to stop the loop
for age in range(12, 24, 2):
    print()
    print("The patient is", age, "years old.")

    if age >= 18:
        break # Stop the loop here

    print("The patient is a minor.")

print()
print("Finished iterating over the ages.")
```

```python
# Using a continue statement to stop the loop (iteration)
for age in range(12, 24, 2):
    print()
    print("The patient is", age, "years old.")

    if age >= 18:
        continue # Stop the current iteration and move on to the next one

    print("The patient is a minor.")

print()
print("Finished iterating over the ages.")
```

```python
# Practice Sheet: Collecting all diseases in a single list.

all_diseases = []

for participant in participants:
    participant_information = participants[participant]
    print(f"{participant}'s diseases are: {participant_information[5]}")
    all_diseases.extend(participant_information[5])

print("All diseases:", all_diseases)
```

```python
# Practice Sheet: Converting the disease list into a set of unique diseases.

unique_diseases = set(all_diseases)

print(unique_diseases)
```

```python
# Practice Sheet: Counting the frequency of each unique disease.

for disease in unique_diseases:
    print(disease, "appears", all_diseases.count(disease), "time(s).")
```

```python
# Practice Sheet: Calculating BMI and classifying each participant.

for participant in participants:
    participant_information = participants[participant]

    gender = participant_information[0]
    height = participant_information[2]
    weight = participant_information[3]

    # Compute the BMI (BMI = weight / height²)

    bmi = weight / (height ** 2)

    if bmi < 18.5:
        category = "Underweight"
    elif bmi <= 25:
        category = "Normal range"
    else:
        category = "Overweight"

    print(f"{participant}, Gender: {gender}, BMI: {bmi:.2f}, Category: {category}")
```

```python
# Practice Sheet: Interactive BMI calculator using a while loop.

while True:
    participant_number = input("Enter participant number (1-10) or 'exit' to stop: ")

    if participant_number == "exit":
        print("Program stopped.")
        break

    key = "participant" + participant_number
    participant_information = participants[key]

    gender = participant_information[0]
    height = participant_information[2]
    weight = participant_information[3]

    bmi = weight / (height ** 2)

    if bmi < 18.5:
        category = "Underweight"
    elif bmi <= 25:
        category = "Normal range"
    else:
        category = "Overweight"

    print(f"Participant {participant_number}")
    print(f"Gender: {gender}")
    print(f"BMI: {bmi:.2f}")
    print(f"Category: {category}")
```
